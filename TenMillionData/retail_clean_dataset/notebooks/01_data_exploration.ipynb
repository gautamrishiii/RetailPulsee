 Build the real daily demand from the full dataset


```python
import pandas as pd

file_path = "../sales_transactions.csv"

chunksize = 200_000
daily_chunks = []

chunk_number = 0

for chunk in pd.read_csv(
    file_path,
    usecols=["date", "quantity"],
    parse_dates=["date"],
    chunksize=chunksize
):
    chunk_number += 1

    daily_chunk = (
        chunk
        .groupby("date")["quantity"]
        .sum()
    )

    daily_chunks.append(daily_chunk)

    if chunk_number % 10 == 0:
        print(f"Processed {chunk_number} chunks...")

# Combine all chunk-level daily totals
full_daily_sales = (
    pd.concat(daily_chunks)
    .groupby(level=0)
    .sum()
    .sort_index()
)

print("\nFinished!")
print("Number of dates:", len(full_daily_sales))
print("Start date:", full_daily_sales.index.min())
print("End date:", full_daily_sales.index.max())
print("\nFirst 10 days:")
print(full_daily_sales.head(10))
```

Create a forecasting function

```python
import pandas as pd
import numpy as np

def forecast_next_days(model, historical_data, days=30):
    
    history = historical_data.copy()
    forecasts = []

    for _ in range(days):
        
        next_date = history.index[-1] + pd.Timedelta(days=1)

        row = {
            "day_of_week": next_date.dayofweek,
            "day_of_month": next_date.day,
            "month": next_date.month,
            "year": next_date.year,
            "day_of_year": next_date.dayofyear,
            "lag_1": history.iloc[-1],
            "lag_7": history.iloc[-7],
            "lag_14": history.iloc[-14],
            "lag_28": history.iloc[-28],
            "rolling_mean_7": history.iloc[-7:].mean(),
            "rolling_mean_30": history.iloc[-30:].mean()
        }

        X_next = pd.DataFrame([row])

        prediction = model.predict(X_next)[0]

        forecasts.append({
            "date": next_date,
            "predicted_demand": prediction
        })

        history.loc[next_date] = prediction

    return pd.DataFrame(forecasts).set_index("date")
```

```python
import pandas as pd

file_path = "../sales_transactions.csv"

chunksize = 200_000
customer_chunks = []

for chunk in pd.read_csv(
    file_path,
    usecols=[
        "date",
        "receipt_id",
        "customer_id",
        "total_value"
    ],
    parse_dates=["date"],
    chunksize=chunksize
):
    customer_chunks.append(chunk)

print("Chunks loaded:", len(customer_chunks))
```

```python
sales_rfm = pd.concat(
    customer_chunks,
    ignore_index=True
)

print("Rows:", len(sales_rfm))
print("Columns:", sales_rfm.columns.tolist())
print(sales_rfm.head())
```

Define the RFM reference date

```python
reference_date = sales_rfm["date"].max() + pd.Timedelta(days=1)

print("Reference date:", reference_date)
print("Last transaction:", sales_rfm["date"].max())
```

Calculate RFM

```python
rfm = sales_rfm.groupby("customer_id").agg(
    Recency=("date", lambda x: (reference_date - x.max()).days),
    Frequency=("receipt_id", "nunique"),
    Monetary=("total_value", "sum")
)

print(rfm.head())
print("\nRFM shape:", rfm.shape)
```

Inspect the RFM data

```python
print(rfm.describe())

print("\nSkewness:")
print(rfm.skew())
```

Log-transform the RFM variables

```python
import numpy as np

rfm_log = np.log1p(rfm)

print(rfm_log.describe())
print("\nSkewness after log transformation:")
print(rfm_log.skew())
```

Standardize the RFM values

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(rfm_log)

rfm_scaled = pd.DataFrame(
    rfm_scaled,
    index=rfm.index,
    columns=rfm.columns
)

print(rfm_scaled.head())
print("\nMeans:")
print(rfm_scaled.mean())

print("\nStandard deviations:")
print(rfm_scaled.std())
```

Find the right number of customer segments

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

K_range = range(2, 9)

for k in K_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(rfm_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(10, 6))

plt.plot(
    K_range,
    inertia,
    marker="o"
)

plt.title("Elbow Method for Customer Segmentation")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.xticks(list(K_range))
plt.grid(True)
plt.show()
```

Calculate silhouette scores

```python
from sklearn.metrics import silhouette_score

silhouette_scores = {}

for k in range(2, 9):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    cluster_labels = kmeans.fit_predict(rfm_scaled)
    
    score = silhouette_score(
        rfm_scaled,
        cluster_labels
    )
    
    silhouette_scores[k] = score

print("Silhouette Scores:")

for k, score in silhouette_scores.items():
    print(f"K={k}: {score:.4f}")
```

```python
inventory_model["demand_status"] = "Active Demand"

inventory_model.loc[
    inventory_model["calendar_avg_daily_demand"].isna(),
    "demand_status"
] = "No Recent Demand"

print(
    inventory_model["demand_status"].value_counts()
)

print("\nNo recent demand records:")
print(
    inventory_model[
        inventory_model["demand_status"] == "No Recent Demand"
    ][
        [
            "store_id",
            "sku_id",
            "sku_name",
            "stock_on_hand",
            "inventory_status"
        ]
    ].head(20).to_string(index=False)
)
```

Calculate SKU demand

```python
inventory_model["store_target_stock"] = (
    inventory_model["calendar_avg_daily_demand"] * 30
    + inventory_model["final_safety_stock"]
)

inventory_model["store_recommended_order_qty"] = (
    inventory_model["store_target_stock"]
    - inventory_model["stock_on_hand"]
).clip(lower=0)

# No automatic reorder when there is no recent demand
inventory_model.loc[
    inventory_model["demand_status"] == "No Recent Demand",
    "store_recommended_order_qty"
] = 0

inventory_model["store_recommended_order_qty"] = (
    inventory_model["store_recommended_order_qty"]
    .round()
    .astype(int)
)

print(
    inventory_model[
        [
            "store_id",
            "sku_id",
            "sku_name",
            "inventory_status",
            "demand_status",
            "stock_on_hand",
            "calendar_avg_daily_demand",
            "final_safety_stock",
            "store_target_stock",
            "store_recommended_order_qty"
        ]
    ]
    .sort_values(
        "store_recommended_order_qty",
        ascending=False
    )
    .head(20)
    .to_string(index=False)
)
```

Calculate store-specific demand variability

```python
store_sku_variability = (
    store_sku_daily
    .groupby(["store_id", "sku_id"])["quantity"]
    .std()
    .reset_index(name="store_demand_std")
)

print("Store-SKU combinations:", len(store_sku_variability))

print("\nSample:")
print(store_sku_variability.head(10))

print("\nMissing variability:")
print(
    store_sku_variability["store_demand_std"].isna().sum()
)
```

Merge the store-specific variability

```python
inventory_model = inventory_model.merge(
    store_sku_variability,
    on=["store_id", "sku_id"],
    how="left"
)

# Use store-SKU variability when available.
# Otherwise fall back to SKU-level variability.
inventory_model["final_demand_std"] = (
    inventory_model["store_demand_std"]
    .fillna(inventory_model["demand_std"])
)

print("Missing final demand variability:")
print(
    inventory_model["final_demand_std"].isna().sum()
)

print("\nSample:")
print(
    inventory_model[
        [
            "store_id",
            "sku_id",
            "calendar_avg_daily_demand",
            "store_demand_std",
            "demand_std",
            "final_demand_std"
        ]
    ].head(10)
)
```

```python
Z = 1.645
lead_time_days = 7

inventory_model["store_statistical_safety_stock"] = (
    Z
    * inventory_model["final_demand_std"]
    * (lead_time_days ** 0.5)
).round().astype(int)

print("Store-specific safety stock summary:")
print(
    inventory_model["store_statistical_safety_stock"].describe()
)

print("\nSample:")
print(
    inventory_model[
        [
            "store_id",
            "sku_id",
            "final_demand_std",
            "store_statistical_safety_stock"
        ]
    ].head(10)
)
```

```python
# Final safety stock:
# Keep the existing business safety stock as the minimum,
# but increase it when statistical safety stock is higher.

inventory_model["store_final_safety_stock"] = (
    inventory_model[
        ["safety_stock", "store_statistical_safety_stock"]
    ].max(axis=1)
)

# 30-day target stock
inventory_model["store_target_stock"] = (
    inventory_model["calendar_avg_daily_demand"] * 30
    + inventory_model["store_final_safety_stock"]
)

# Recommended replenishment quantity
inventory_model["store_recommended_order_qty"] = (
    inventory_model["store_target_stock"]
    - inventory_model["stock_on_hand"]
).clip(lower=0)

# Do not automatically reorder products with no recent demand
inventory_model.loc[
    inventory_model["demand_status"] == "No Recent Demand",
    "store_recommended_order_qty"
] = 0

inventory_model["store_recommended_order_qty"] = (
    inventory_model["store_recommended_order_qty"]
    .round()
    .astype(int)
)

print("Final inventory model created.")

print("\nRecommended order quantity summary:")
print(
    inventory_model["store_recommended_order_qty"].describe()
)

print("\nTotal recommended units:")
print(
    inventory_model["store_recommended_order_qty"].sum()
)
```

```python
top_reorders = (
    inventory_model[
        inventory_model["store_recommended_order_qty"] > 0
    ]
    .sort_values(
        "store_recommended_order_qty",
        ascending=False
    )
    .head(20)
)

print(
    top_reorders[
        [
            "store_id",
            "sku_id",
            "sku_name",
            "category",
            "stock_on_hand",
            "calendar_avg_daily_demand",
            "store_final_safety_stock",
            "store_target_stock",
            "store_recommended_order_qty",
            "demand_status"
        ]
    ].to_string(index=False)
)
```

```python
# Store-specific days of stock
inventory_model["store_days_of_stock"] = (
    inventory_model["stock_on_hand"]
    / inventory_model["calendar_avg_daily_demand"]
)

# No recent demand -> days of stock is not meaningful
inventory_model.loc[
    inventory_model["demand_status"] == "No Recent Demand",
    "store_days_of_stock"
] = np.nan


def classify_store_inventory(row):
    if row["demand_status"] == "No Recent Demand":
        return "No Recent Demand"
    
    days = row["store_days_of_stock"]
    
    if days == 0:
        return "Stockout"
    elif days < 7:
        return "Critical"
    elif days < 14:
        return "Low Stock"
    elif days < 30:
        return "Moderate"
    elif days <= 180:
        return "Healthy"
    else:
        return "Overstock"


inventory_model["final_inventory_status"] = (
    inventory_model.apply(classify_store_inventory, axis=1)
)

print(
    inventory_model["final_inventory_status"]
    .value_counts()
)
```

```python
inventory_value_summary = (
    inventory_model
    .assign(
        inventory_value=lambda x:
            x["stock_on_hand"] * x["cost_price"]
    )
    .groupby("final_inventory_status")
    .agg(
        records=("sku_id", "count"),
        units_on_hand=("stock_on_hand", "sum"),
        inventory_value=("inventory_value", "sum"),
        recommended_units=("store_recommended_order_qty", "sum")
    )
    .sort_values("inventory_value", ascending=False)
)

print(inventory_value_summary)
```

```python
total_records = len(inventory_model)

stockout_records = (
    inventory_model["final_inventory_status"] == "Stockout"
).sum()

critical_records = (
    inventory_model["final_inventory_status"] == "Critical"
).sum()

low_stock_records = (
    inventory_model["final_inventory_status"] == "Low Stock"
).sum()

overstock_records = (
    inventory_model["final_inventory_status"] == "Overstock"
).sum()

no_demand_records = (
    inventory_model["final_inventory_status"] == "No Recent Demand"
).sum()

total_inventory_units = inventory_model["stock_on_hand"].sum()

total_inventory_value = (
    inventory_model["stock_on_hand"]
    * inventory_model["cost_price"]
).sum()

total_recommended_units = (
    inventory_model["store_recommended_order_qty"].sum()
)

stockout_recommended_units = (
    inventory_model.loc[
        inventory_model["final_inventory_status"] == "Stockout",
        "store_recommended_order_qty"
    ].sum()
)

overstock_inventory_value = (
    inventory_model.loc[
        inventory_model["final_inventory_status"] == "Overstock",
        "stock_on_hand"
    ]
    * inventory_model.loc[
        inventory_model["final_inventory_status"] == "Overstock",
        "cost_price"
    ]
).sum()

print("========== RETAILPULSE INVENTORY KPIs ==========")

print(f"Total inventory records: {total_records:,}")
print(f"Total units on hand: {total_inventory_units:,}")
print(f"Total inventory value at cost: {total_inventory_value:,.2f}")

print(f"\nStockout records: {stockout_records:,}")
print(f"Critical records: {critical_records:,}")
print(f"Low-stock records: {low_stock_records:,}")
print(f"Overstock records: {overstock_records:,}")
print(f"No-recent-demand records: {no_demand_records:,}")

print(f"\nTotal recommended replenishment: {total_recommended_units:,} units")
print(f"Stockout replenishment: {stockout_recommended_units:,} units")

print(
    f"\nOverstock inventory value: "
    f"{overstock_inventory_value:,.2f}"
)

print(
    f"\nStockout share of recommended units: "
    f"{stockout_recommended_units / total_recommended_units * 100:.2f}%"
)
```
